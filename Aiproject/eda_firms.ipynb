```python
fires_by_date = df.groupby('acq_date').size()
plt.figure(figsize=(10, 5))
plt.bar(fires_by_date.index.strftime('%b %d'), fires_by_date.values)
plt.xlabel('Date')
plt.ylabel('Number of FIRMS Detections')
plt.title('Daily NASA FIRMS Fire Detections')
plt.tight_layout()
plt.show()
```

```python
numeric_columns = ['latitude', 'longitude', 'bright_ti4', 'bright_ti5', 'scan', 'track', 'frp', 'log_frp']
numeric_columns = [col for col in numeric_columns if col in df.columns]
correlation = df[numeric_columns].corr()
correlation
```

```python
plt.figure(figsize=(12, 9))
sns.heatmap(correlation, annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Correlation Matrix of FIRMS Variables')
plt.tight_layout()
plt.show()
```

```python
frp_threshold = df['frp'].quantile(0.99)
top_frp = df[df['frp'] >= frp_threshold]
top_frp[['latitude', 'longitude', 'acq_date', 'frp', 'bright_ti4', 'confidence']].sort_values('frp', ascending=False).head(20)
```

```python
plt.figure(figsize=(11, 8))
plt.scatter(df['longitude'], df['latitude'], s=8, alpha=0.2)
plt.scatter(top_frp['longitude'], top_frp['latitude'], s=30, alpha=0.8)
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.title('Locations of Highest-Intensity FIRMS Detections')
plt.tight_layout()
plt.show()
```
