```python
import xarray as xr
import pandas as pd
import numpy as np
```

```python
instant_file = 'data_stream-oper_stepType-instant.nc'
precip_file = 'data_stream-oper_stepType-accum.nc'
ds_instant = xr.open_dataset(instant_file, engine="netcdf4")
ds_precip = xr.open_dataset(precip_file, engine="netcdf4")
```

```python
df_instant = ds_instant[['t2m', 'd2m', 'u10', 'v10', 'sp']].to_dataframe().reset_index()
df_precip = ds_precip[['tp']].to_dataframe().reset_index()
```

```python
df = pd.merge(df_instant, df_precip, on=['valid_time', 'latitude', 'longitude'], how='inner')
```

```python
df.rename(columns={'valid_time': 'datetime', 'sp': 'surface_pressure'}, inplace=True)
df['datetime'] = pd.to_datetime(df['datetime'])
```

```python
df['temperature_c'] = df['t2m'] - 273.15
df['dewpoint_c'] = df['d2m'] - 273.15
df['wind_speed'] = np.sqrt(df['u10'] ** 2 + df['v10'] ** 2)
df['precipitation_mm'] = df['tp'] * 1000
```

```python
df['date'] = df['datetime'].dt.date
df['hour'] = df['datetime'].dt.hour
df['day'] = df['datetime'].dt.day
df['month'] = df['datetime'].dt.month
df['day_of_week'] = df['datetime'].dt.dayofweek
```

```python
df.drop_duplicates(subset=['datetime', 'latitude', 'longitude'], inplace=True)
```
